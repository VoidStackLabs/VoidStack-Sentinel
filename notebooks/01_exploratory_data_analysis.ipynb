```python
plt.figure(figsize=(6, 5))

sns.countplot(
    data=df,
    x="Class"
)

plt.title("Legitimate vs Fraudulent Transactions")
plt.xlabel("Transaction Class")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="Amount",
    hue="Class",
    bins=50,
    log_scale=(False, True),
    element="step"
)

plt.title("Transaction Amount Distribution")
plt.show()
```

2. Transaction Distribution Analysis

```python
# Basic Statistics for transactional amount
df.groupby("Class")["Amount"].describe()
```

```python
#Comparing the amount distribution into 2 classes
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="Class",
    y="Amount"
)

plt.title("Transaction Amount Distribution by Class")
plt.xlabel("Transaction Class")
plt.ylabel("Transaction Amount")

plt.show()
```

3. Analyzing feature correlations

```python
correlation_matrix = df.corr()
correlation_matrix.T
```

```python
plt.figure(figsize=(14, 14))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Feature Correlation Matrix")

plt.show()
```

```python
class_correlation = (
    correlation_matrix["Class"]
    .sort_values(ascending=False)
)

print(class_correlation)
```

```python
#Five features with the highest positive correlation with Class is V11,V4,V2,V21,V19
#Five features with the highest negative correlation with Class is V17,V14,V12,V7,V18
#Whether Amount has a strong or weak linear correlation with Class:Looking at the Amount row/column,
#The correlations with all the V features are very faint (light pink/blue, close to white). 
#The cell at the intersection of Amount and Class is similarly light, indicating a weak linear correlation.


#Correlation does not establish causation, and weak correlation does not mean 
# a feature is useless for a nonlinear model.
```

4. Comparing legitimate and fraudulent transactions

```python
amount_analysis = df.groupby("Class").agg(
    transaction_count=("Amount", "count"),
    average_amount=("Amount", "mean"),
    median_amount=("Amount", "median"),
    maximum_amount=("Amount", "max"),
    minimum_amount=("Amount", "min")
)

display(amount_analysis)
```
